```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("iris.csv")
```

```python
df.head()
```

```python
sns.set(style="whitegrid")
df.head()
```

```python
df.hist(figsize=(10,10), bins=20, edgecolor="black")
plt.suptitle("Histogram of Iris Dataset Features")
plt.tight_layout()
```

```python
plt.figure(figsize=(10,10))
sns.boxplot(data=df.drop(columns="species"))
```

```python
plt.figure(figsize=(10,10))
df.groupby("species").mean().T.plot(marker='o')
plt.xlabel("Species")
plt.legend(title="Species")
plt.ylabel("Mean Values")
```

```python
plt.figure(figsize=(10, 6))
plt.plot(df["sepal_length"], label="Sepal Length")
plt.plot(df["sepal_width"], label="Sepal Width")
plt.plot(df["petal_length"], label="Petal Length")
plt.plot(df["petal_width"], label="Petal Width")
plt.title("Multiple Line Graph of Features")
plt.xlabel("Index")
plt.ylabel("Measurement")
plt.legend()
plt.grid(True)
plt.show()
```
