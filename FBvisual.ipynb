```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('dataset_Facebook.csv', sep=';', nrows=100)
```

```python
df.columns
```

```python
plt.figure(figsize=(20,10))
sns.boxplot(data=df[['comment', 'like', 'share']])
plt.tight_layout()
```

```python
#histo
plt.figure(figsize=(15,10))
plt.hist(df['Total Interactions'], bins=40, color='skyblue', edgecolor='black')
plt.tight_layout()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df.index,df['like'],color='blue',marker='o')
plt.grid(True)
```

```python
plt.figure(figsize=(12,12))
plt.title("Multiline Graph")
plt.plot(df.index,df['like'],color="blue",marker='o')
plt.plot(df.index,df['comment'],color="green",marker='s')
plt.plot(df.index,df['share'],color="red",marker='^')
plt.legend()
plt.grid(True)
```
