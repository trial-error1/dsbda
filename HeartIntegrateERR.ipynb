```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
heart = pd.read_csv('heart.csv')
```

```python
heart.fillna(heart.mode().iloc[0], inplace=True)
```

```python
heart.drop_duplicates(inplace=True)
```

```python
heart.shape
```

```python
print("Data Cleaning Done")
```

```python
heart.columns
```

```python
df1 = heart[['age', 'sex', 'cp', 'trestbps']]
df2 = heart[['trestbps', 'chol', 'fbs', 'restecg']]
```

```python
data = pd.merge(df1,df2,on='trestbps',how='inner')
data.head()
```

```python
#error correction
if 'age' in heart.columns:
    heart = heart[(heart['age'] > 0) & (heart['age'] < 120)]
```

```python
if 'chol' in heart.columns:
    heart = heart[heart['chol'] > 0]
```

```python
heart.head()
```

```python
#model building

x = heart.drop(columns=['age'])
y = heart['cp']
y = y.round().astype(int)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
model = RandomForestClassifier()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
#accuracy
accuracy = accuracy_score(y_test,y_pred)
print(accuracy)
```
