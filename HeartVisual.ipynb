```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('heart.csv')
```

```python
df.columns
```

```python
plt.figure(figsize=(10,10))
sns.boxplot(data=df[['age','chol','thalach','oldpeak']])
```

```python
df[['age','chol','thalach','oldpeak']].hist(figsize=(10,10),bins=20,color='blue',edgecolor='black')
```

```python
sns.lineplot(data=df,x = df['age'],y = df['chol'],color='red',marker='o')
```

```python
df_sorted = df.sort_values(by='age').iloc[:100]
```

```python
plt.figure(figsize=(15,15))
plt.plot(df_sorted['age'],df_sorted['thalach'], label='Max Heart Rate', marker='o')
plt.plot(df_sorted['age'],df_sorted['chol'],label = 'Cholestrol', marker='s')
plt.plot(df_sorted['age'],df_sorted['oldpeak'], label="OldPeak", marker='^')
```
